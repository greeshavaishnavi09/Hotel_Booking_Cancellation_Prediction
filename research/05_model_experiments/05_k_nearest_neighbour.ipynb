```python
# import current working directory

import os

print(os.getcwd())
```

```python
# import Libraries

import pandas as pd
import numpy as np

from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    confusion_matrix,
    f1_score,
    roc_auc_score,
    average_precision_score,
    log_loss
)
```

```python
# scaled transformed data will be used for svm

train_df = pd.read_csv(
    "../../artifacts/data_transformation/train.csv"
)

test_df = pd.read_csv(
    "../../artifacts/data_transformation/test.csv"
)

print("Train shape:", train_df.shape)
print("Test shape :", test_df.shape)
```

```python
# seperate features and target 

X_train = train_df.drop("is_canceled", axis=1)

y_train = train_df["is_canceled"]

X_test = test_df.drop( "is_canceled",axis=1)

y_test = test_df["is_canceled"]
```

```python
# checking shapes 

print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("X_test shape :", X_test.shape)
print("y_test shape :", y_test.shape)
```

```python
# checking target distribution

print("Training target distribution:")
print(y_train.value_counts())

print("\nTraining target percentage:")
print(y_train.value_counts(normalize=True) * 100)
```

# KNN Baseline model

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
```

```python
# train the model

knn_model.fit(X_train,
    y_train
)
```

```python
# prediction

knn_pred = knn_model.predict(X_test)

knn_prob = knn_model.predict_proba(X_test)[:, 1]
```

```python
# evaluation 

knn_cm = confusion_matrix(
    y_test,
    knn_pred
)

tn, fp, fn, tp = knn_cm.ravel()


knn_accuracy = accuracy_score(
    y_test,
    knn_pred
)

knn_precision = precision_score(
    y_test,
    knn_pred,
    zero_division=0
)

knn_recall = recall_score(
    y_test,
    knn_pred,
    zero_division=0
)

knn_specificity = tn / (tn + fp)

knn_f1 = f1_score(
    y_test,
    knn_pred,
    zero_division=0
)

knn_roc_auc = roc_auc_score(
    y_test,
    knn_prob
)

knn_pr_auc = average_precision_score(
    y_test,
    knn_prob
)

knn_logloss = log_loss(
    y_test,
    knn_prob
)
```

```python
# display result

print("=" * 55)
print("KNN - BASELINE EVALUATION")
print("=" * 55)

print(f"Accuracy    : {knn_accuracy:.4f}")
print(f"Precision   : {knn_precision:.4f}")
print(f"Recall      : {knn_recall:.4f}")
print(f"Specificity : {knn_specificity:.4f}")
print(f"F1 Score    : {knn_f1:.4f}")
print(f"ROC-AUC     : {knn_roc_auc:.4f}")
print(f"PR-AUC      : {knn_pr_auc:.4f}")
print(f"Log Loss    : {knn_logloss:.4f}")
```

```python
knn_baseline_results = pd.DataFrame({
    "Model": ["KNN"],
    "Accuracy": [knn_accuracy],
    "Precision": [knn_precision],
    "Recall": [knn_recall],
    "Specificity": [knn_specificity],
    "F1 Score": [knn_f1],
    "ROC-AUC": [knn_roc_auc],
    "PR-AUC": [knn_pr_auc],
    "Log Loss": [knn_logloss]
})

knn_baseline_results.round(4)
```

```python
# KNN n_neighbour experiment 

k_values = [3, 5, 7, 11]

knn_k_results = []

for k in k_values:

    model = KNeighborsClassifier(
        n_neighbors=k,
        n_jobs=-1
    )

    model.fit(
        X_train,
        y_train
    )

    test_pred = model.predict(
        X_test
    )

    test_prob = model.predict_proba(
        X_test
    )[:, 1]

    cm = confusion_matrix(
        y_test,
        test_pred
    )

    tn, fp, fn, tp = cm.ravel()

    accuracy = accuracy_score(
        y_test,
        test_pred
    )

    precision = precision_score(
        y_test,
        test_pred,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        test_pred,
        zero_division=0
    )

    specificity = tn / (tn + fp)

    f1 = f1_score(
        y_test,
        test_pred,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        test_prob
    )

    pr_auc = average_precision_score(
        y_test,
        test_prob
    )

    logloss = log_loss(
        y_test,
        test_prob
    )

    knn_k_results.append({
        "K": k,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "Specificity": specificity,
        "F1 Score": f1,
        "ROC-AUC": roc_auc,
        "PR-AUC": pr_auc,
        "Log Loss": logloss
    })

    
```

```python
# creating dataframe 

knn_k_results_df = pd.DataFrame(
    knn_k_results
)

knn_k_results_df.round(4)
```

# hyper parameter tuning

```python
# import gridsearch cv

from sklearn.model_selection import GridSearchCV
```

```python
# parameter grid

knn_param_grid = {
    "n_neighbors": [5, 7, 11],
    "weights": ["uniform", "distance"]
}
```

```python
# create gridsearch cv 

knn_grid_search = GridSearchCV(
    estimator=KNeighborsClassifier(
        n_jobs=-1
    ),
    param_grid=knn_param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1,
    verbose=1
)
```

```python
# model fit 

knn_grid_search.fit(X_train,y_train)
```

```python
# get the best parameter

print("Best Parameters:")
print(knn_grid_search.best_params_)

print("\nBest CV F1 Score:")
print(knn_grid_search.best_score_)
```

```python
# estimator from gridsearchcv

knn_tuned = knn_grid_search.best_estimator_
```

```python
# make the prediction and probability of tunned model

knn_tuned_pred = knn_tuned.predict(
    X_test
)

knn_tuned_prob = knn_tuned.predict_proba(
    X_test
)[:, 1]
```

```python
knn_tuned_cm = confusion_matrix(
    y_test,
    knn_tuned_pred
)

tn, fp, fn, tp = knn_tuned_cm.ravel()


knn_tuned_accuracy = accuracy_score(
    y_test,
    knn_tuned_pred
)

knn_tuned_precision = precision_score(
    y_test,
    knn_tuned_pred,
    zero_division=0
)

knn_tuned_recall = recall_score(
    y_test,
    knn_tuned_pred,
    zero_division=0
)

knn_tuned_specificity = tn / (tn + fp)

knn_tuned_f1 = f1_score(
    y_test,
    knn_tuned_pred,
    zero_division=0
)

knn_tuned_roc_auc = roc_auc_score(
    y_test,
    knn_tuned_prob
)

knn_tuned_pr_auc = average_precision_score(
    y_test,
    knn_tuned_prob
)

knn_tuned_logloss = log_loss(
    y_test,
    knn_tuned_prob
)
```

```python
# display result

print("=" * 55)
print("KNN - TUNED MODEL EVALUATION")
print("=" * 55)

print(f"Accuracy    : {knn_tuned_accuracy:.4f}")
print(f"Precision   : {knn_tuned_precision:.4f}")
print(f"Recall      : {knn_tuned_recall:.4f}")
print(f"Specificity : {knn_tuned_specificity:.4f}")
print(f"F1 Score    : {knn_tuned_f1:.4f}")
print(f"ROC-AUC     : {knn_tuned_roc_auc:.4f}")
print(f"PR-AUC      : {knn_tuned_pr_auc:.4f}")
print(f"Log Loss    : {knn_tuned_logloss:.4f}")
```

```python
# baseline vs tuned model compare

knn_comparison = pd.DataFrame({
    "Metric": [
        "Accuracy",
        "Precision",
        "Recall",
        "Specificity",
        "F1 Score",
        "ROC-AUC",
        "PR-AUC",
        "Log Loss"
    ],

    "Baseline": [
        knn_accuracy,
        knn_precision,
        knn_recall,
        knn_specificity,
        knn_f1,
        knn_roc_auc,
        knn_pr_auc,
        knn_logloss
    ],

    "Tuned": [
        knn_tuned_accuracy,
        knn_tuned_precision,
        knn_tuned_recall,
        knn_tuned_specificity,
        knn_tuned_f1,
        knn_tuned_roc_auc,
        knn_tuned_pr_auc,
        knn_tuned_logloss
    ]
})

knn_comparison.round(4)
```
